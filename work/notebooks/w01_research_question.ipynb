import pandas as pd

# Load the public UCI Heart Disease dataset directly from the web
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"
cols = ['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'target']
df = pd.read_csv(url, names=cols)

# Convert target to a strict binary classification (0 = no disease, 1 = disease present)
df['target'] = df['target'].apply(lambda x: 1 if x > 0 else 0)

## 2) The question: decision, action, cost of a wrong call
* **Decision:** Does this patient have a high probability of heart disease requiring immediate further screening?
* **Action:** Flag the patient's profile for priority review by a cardiologist and schedule a follow-up stress test.
* **Cost of a wrong call:** 
    * *False Positive:* The patient experiences unnecessary anxiety, and the hospital incurs the financial cost and time of unneeded tests.
    * *False Negative (Worst Case):* A high-risk patient is sent home without intervention, potentially leading to severe, life-threatening health complications.
* **Why ML:** A simple fixed rule (e.g., "Flag everyone with cholesterol over 250") ignores the complex, non-linear interactions between multiple risk factors like age, max heart rate, and chest pain type. ML can weigh all these interacting features simultaneously to make a highly nuanced prediction.
